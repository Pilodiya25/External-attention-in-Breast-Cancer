```python
import pandas as pd
import numpy as np
```

```python
labels = pd.read_csv("labels_one_hot_encoded.csv")
gene_expression = pd.read_csv("mobrca_gene.csv", index_col=0)
methylation = pd.read_csv("mobrca_methylation.csv", index_col=0)
mirna = pd.read_csv("mobrca_mirna.csv", index_col=0)

# Display shape of each dataset
print("Labels shape:", labels.shape)
print("Gene expression shape:", gene_expression.shape)
print("Methylation shape:", methylation.shape)
print("miRNA shape:", mirna.shape)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy
```

```python
# Step 2: Separate features and labels for gene expression
X_gene = gene_expression.values
y = labels.values.argmax(axis=1)  # Convert one-hot to class labels

# Step 3: Initialize RandomForest and Boruta
rf_gene = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
boruta_gene = BorutaPy(estimator=rf_gene, n_estimators='auto', random_state=42)

# Step 4: Fit Boruta
print("Running Boruta on gene expression data...")
boruta_gene.fit(X_gene, y)

# Step 5: Get selected features
selected_gene_features = gene_expression.columns[boruta_gene.support_].tolist()

# Step 6: Create a new DataFrame with selected features
gene_selected = gene_expression[selected_gene_features]

# Print results
print(f"Selected {len(selected_gene_features)} features from gene expression data.")
```

```python
gene_selected.to_csv("gene_selected.csv", index=False)
```

```python
# Step 2: Separate features and labels for methylation
X_methylation = methylation.values
y = labels.values.argmax(axis=1)  # Convert one-hot to class labels

# Step 3: Initialize RandomForest and Boruta
rf_methylation = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
boruta_methylation = BorutaPy(estimator=rf_methylation, n_estimators='auto', random_state=42)

# Step 4: Fit Boruta
print("Running Boruta on methylation data...")
boruta_methylation.fit(X_methylation, y)

# Step 5: Get selected features
selected_methylation_features = methylation.columns[boruta_methylation.support_].tolist()

# Step 6: Create a new DataFrame with selected features
methylation_selected = methylation[selected_methylation_features]

# Print results
print(f"Selected {len(selected_methylation_features)} features from methylation data.")
```

```python
methylation_selected.to_csv("methylation_selected.csv", index=False)
```

```python
# Step 2: Separate features and labels for mirna
X_mirna = mirna.values
y = labels.values.argmax(axis=1)  # Convert one-hot to class labels

# Step 3: Initialize RandomForest and Boruta
rf_mirna = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
boruta_mirna = BorutaPy(estimator=rf_mirna, n_estimators='auto', random_state=42)

# Step 4: Fit Boruta
print("Running Boruta on mirna data...")
boruta_mirna.fit(X_mirna, y)

# Step 5: Get selected features
selected_mirna_features = mirna.columns[boruta_mirna.support_].tolist()

# Step 6: Create a new DataFrame with selected features
mirna_selected = mirna[selected_mirna_features]

# Print results
print(f"Selected {len(selected_mirna_features)} features from mirnan data.")
```

```python
mirna_selected.to_csv("mirna_selected.csv", index=False)
```

```python
# Concatenate the selected features
combined_features = pd.concat(
    [gene_expression[selected_gene_features],
     methylation[selected_methylation_features],
     mirna[selected_mirna_features]],
    axis=1
)

print("Combined feature shape:", combined_features.shape)
```

```python
# Confirming shapes before moving forward
print(f"Combined features shape: {combined_features.shape}")
print(f"Labels shape: {labels.shape}")
```

```python
combined_features.to_csv("combined_features.csv", index=False)
```
